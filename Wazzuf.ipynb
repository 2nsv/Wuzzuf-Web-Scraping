# Scrapping

```python
import selenium
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time

 
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")


Positions = ["Data Engineer", "Data Science", "AI Developer", "Machine Learning Engineer"]
Jobs = []

driver = webdriver.Chrome(options=options)
driver.get("https://wuzzuf.net/")


for Position in Positions :    
    driver.find_element(By.NAME ,"q").clear()
    driver.find_element(By.NAME,'q').send_keys(Position)
    driver.find_element(By.XPATH,"//button[@type='submit']").click()
    time.sleep(2)
    driver.refresh()
    time.sleep(5)
    for Page in range (4):
        time.sleep(5)
        job_scraper = driver.find_elements(By.CSS_SELECTOR,".css-ghe2tq")
        time.sleep(5)
        for i in job_scraper :
            Job_Title = i.find_element(By.CLASS_NAME,'css-s5fwzh').text
            Company = i.find_element(By.CLASS_NAME,'css-ipsyv7').text
            Location = i.find_element(By.CLASS_NAME,'css-16x61xq').text
            try :
                Experience = i.find_element(By.XPATH, ".//span[contains(., 'Yrs of Exp')]").text.strip(" ·")
            except :
                Experience = 'No Experince Found'    
            Job_URL = i.find_element(By.CLASS_NAME,'css-o171kl').get_attribute('href')
            Jobs.append({
            "Job_Title" : Job_Title,
            "Company" : Company,
            "Location" : Location,
            "Experience" : Experience,
            "Job_URL" : Job_URL
            })
        try :
         driver.find_element(By.CSS_SELECTOR,"a[aria-label='Next page'] i[class='css-b80bsd efou2fk0'] svg").click()
        except :
         continue    
```

# Small Transformation & File Save

```python
import pandas as pd

Final = pd.DataFrame(Jobs)
```
